```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_CANDIDATES = [Path("data/tips.csv"), Path("../data/tips.csv")]
DATA_PATH = next((path for path in DATA_CANDIDATES if path.exists()), None)
DATA_SOURCE = DATA_PATH if DATA_PATH is not None else "https://raw.githubusercontent.com/Eshankulov/cs-31-24-Adenov/main/data/tips.csv"
df = pd.read_csv(DATA_SOURCE)
print("Размер таблицы:", df.shape)
df.head()
```

```python
df.info()
print("\nПропуски по столбцам:")
print(df.isna().sum())
print("\nБазовые статистики:")
display(df.describe(include="all").T)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df["total_bill"], bins=20, edgecolor="black")
axes[0].set(title="Распределение общего счёта", xlabel="Счёт", ylabel="Количество")
df["day"].value_counts().reindex(["Thur", "Fri", "Sat", "Sun"]).plot(kind="bar", ax=axes[1])
axes[1].set(title="Число посещений по дням", xlabel="День", ylabel="Количество")
plt.tight_layout(); plt.show()
```
